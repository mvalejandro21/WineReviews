```python
# Ver los 10 vinos más caros
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('../data/raw-winemag-data-130k-v2.csv')
df.sort_values('price', ascending=False)[['title', 'country', 'points', 'variety', 'winery', 'price']].head(20)
```

```python


# Filtrar vinos con EXACTAMENTE 90 puntos
vinos_90_puntos = df[df['points'] == 90]

# Seleccionar solo las columnas necesarias
resultado = vinos_90_puntos[['title', 'points', 'price', 'country', 'province', 'variety']]

# Tomar solo los primeros 10 vinos (sin ordenar por precio)
resultado = resultado.head(10)

# Mostrar resultados
print("🍷 10 VINOS CON 90 PUNTOS")
print("=" * 80)

for index, row in resultado.iterrows():
    print(f"🍇 {row['title']}")
    print(f"   📊 Puntos: {row['points']} | 💰 Precio: ${row['price']}")
    print(f"   🌍 País: {row['country']} | 📍 Región: {row['province']}")
    print(f"   🍷 Variedad: {row['variety']}")
    print("-" * 80)

# Opcional: Guardar resultados en CSV
resultado.to_csv('10_vinos_90_puntos.csv', index=False)
print(f"\n💾 Resultados guardados en '10_vinos_90_puntos.csv'")

# Estadísticas básicas
print(f"\n📈 RESUMEN:")
print(f"Rango de precios: ${resultado['price'].min()} - ${resultado['price'].max()}")
print(f"Precio promedio: ${resultado['price'].mean():.2f}")
print(f"Países representados: {', '.join(resultado['country'].unique())}")
```
